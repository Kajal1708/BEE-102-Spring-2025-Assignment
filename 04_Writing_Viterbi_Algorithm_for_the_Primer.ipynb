```python
#viterbi
import numpy as np
import math

# States and symbols
states = ['E', 'I']
symbols = ['A', 'C', 'G', 'T']

# Transition probabilities (Nature Primer example)
#        To
# From   E     I
trans_probs = {
    'E': {'E': 0.9, 'I': 0.1},
    'I': {'E': 0.1, 'I': 0.9}
}

# Emission probabilities
# P(symbol | state)
emit_probs = {
    'E': {'A': 0.25, 'C': 0.25, 'G': 0.25, 'T': 0.25},
    'I': {'A': 0.4,  'C': 0.1,  'G': 0.1,  'T': 0.4}
}

# Initial probabilities
start_probs = {
    'E': 0.5,
    'I': 0.5
}

# -----------------------------
# Part A: Log probability of a given path
# -----------------------------
def get_log_prob_of_a_given_path(path, sequence):
    log_prob = math.log(start_probs[path[0]]) + math.log(emit_probs[path[0]][sequence[0]])
    for i in range(1, len(path)):
        prev_state = path[i - 1]
        curr_state = path[i]
        obs = sequence[i]
        log_prob += math.log(trans_probs[prev_state][curr_state])
        log_prob += math.log(emit_probs[curr_state][obs])
    return round(log_prob, 2)

# Example usage:
path = "EEEEEEEEEEEEEEEEEIIIIIII"
obs = "CTTCATGTGAAAGCAGACGTAAGTCA"
print("Log-prob of path:", get_log_prob_of_a_given_path(path, obs))

def viterbi(obs_seq):
    n = len(obs_seq)
    dp = [{} for _ in range(n)]
    backpointer = [{} for _ in range(n)]

    # Initialization
    for s in states:
        dp[0][s] = math.log(start_probs[s]) + math.log(emit_probs[s][obs_seq[0]])
        backpointer[0][s] = None

    # Recursion
    for t in range(1, n):
        for curr in states:
            max_prob, prev_state = max(
                (dp[t - 1][prev] + math.log(trans_probs[prev][curr]) + math.log(emit_probs[curr][obs_seq[t]]), prev)
                for prev in states
            )
            dp[t][curr] = max_prob
            backpointer[t][curr] = prev_state

    # Termination
    last_state = max(states, key=lambda s: dp[-1][s])
    best_path = [last_state]
    for t in range(n - 1, 0, -1):
        last_state = backpointer[t][last_state]
        best_path.append(last_state)
    best_path.reverse()

    return ''.join(best_path), round(dp[-1][best_path[-1]], 2)

# Example usage:
viterbi_path, viterbi_log_prob = viterbi(obs)
print("Viterbi most likely path:", viterbi_path)
print("Log-prob of Viterbi path:", viterbi_log_prob)
```
